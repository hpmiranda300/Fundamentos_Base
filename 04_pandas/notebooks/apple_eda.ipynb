```python
# Convertendo uma série de datas para o formato datetime
datas = pd.to_datetime(
    df_apple["data"],
    format="%Y-%m-%d",
    errors="coerce"
)

# Exibindo as datas convertidas de str para datetime
datas

#O que acontece:
#errors="coerce": transforma valores inválidos em NaT (Not a Time).
#Datas válidas são convertidas para datetime64[ns].

#O parâmetro format dentro do pd.to_datetime() não define como você quer que a data fique, e sim como a data original está escrita no texto 
# para o Pandas conseguir lê-la.
#
#A sua data original é: "2020-09-19" (Ano-Mês-Dia → %Y-%m-%d).
#Ao colocar format="%d-%m-%Y", você está dizendo: "Pandas, procure uma data onde os primeiros números sejam o dia".
#O Pandas tenta ler "2020" como dia, não consegue e, como você usou errors="coerce", ele silenciosamente converte a linha para NaT (Not a Time) 
# em vez de travar o código.
```

```python
#Como fazer o que você quer?
#
#A regra de ouro no Pandas é: primeiro você lê a data no formato em que ela existe, e depois você formata para o formato que você quer.
#
#Opção 1: Em duas etapas (mais fácil de entender)

# 1. Lê a data no formato em que ela realmente está (Ano-Mês-Dia)
datas_convertidas = pd.to_datetime(df_apple["data"], format="%Y-%m-%d", errors="coerce")

# 2. Agora sim transforma para Dia-Mês-Ano usando .dt.strftime()
datas_convertidas = datas_convertidas.dt.strftime("%d/%m/%Y")

datas_convertidas

#Opção 2: Em uma linha só
#datas_convertidas = pd.to_datetime(df_apple["data"]).dt.strftime("%d-%m-%Y")
#datas_convertidas

#(Se preferir com barras como no padrão brasileiro, basta trocar por "%d/%m/%Y").
```

```python
# Criando um DataFrame para comparar as datas originais e convertidas
comparacoes = pd.DataFrame({
    "data_original": datas,
    "data_convertida": datas_convertidas
})

# Exibindo as comparações
comparacoes
```
